```python
df = df.drop(columns=['weight'])
cols = ['payer_code', 'medical_specialty', 'race', 'diag_1', 'diag_2', 'diag_3']
df[cols] = df[cols].fillna('Unknown')
print(df.isna().sum().sum())   # should print 0
```

```python
from sklearn.model_selection import GroupShuffleSplit

print('readmitted in df:', 'readmitted' in df.columns)

X = df.drop(columns=['target', 'patient_nbr', 'readmitted'], errors='ignore')
y = df['target']
groups = df['patient_nbr']

gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, test_idx = next(gss.split(X, y, groups=groups))
X_train, X_test = X.iloc[train_idx].copy(), X.iloc[test_idx].copy()
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

assert 'readmitted' not in X_train.columns, "readmitted is still in X_train"
print('X_train columns:', X_train.shape[1])

# Checks
print(X_train.shape, X_test.shape)
print(y_train.mean(), y_test.mean())
train_patients = set(df.iloc[train_idx]['patient_nbr'])
test_patients = set(df.iloc[test_idx]['patient_nbr'])
print(len(train_patients & test_patients))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (average_precision_score, roc_auc_score,
                             precision_score, recall_score, classification_report)

# 1. Numeric-looking IDs are really categories, so make them strings
id_cols = ['admission_type_id', 'discharge_disposition_id', 'admission_source_id']
X_train = X_train.copy()
X_test = X_test.copy()
X_train[id_cols] = X_train[id_cols].astype(str)
X_test[id_cols] = X_test[id_cols].astype(str)

# 2. Split columns by type
num_cols = ['time_in_hospital', 'num_lab_procedures', 'num_procedures',
            'num_medications', 'number_outpatient', 'number_emergency',
            'number_inpatient', 'number_diagnoses']
cat_cols = [c for c in X_train.columns if c not in num_cols]

# 3. Preprocessing: scale numbers, one-hot encode categories
preprocessor = ColumnTransformer([
    ('num', StandardScaler(), num_cols),
    ('cat', OneHotEncoder(handle_unknown='infrequent_if_exist',
                          min_frequency=50), cat_cols),
])

# 4. Pipeline = preprocessing + model
model = Pipeline([
    ('prep', preprocessor),
    ('clf', LogisticRegression(class_weight='balanced', max_iter=1000)),
])

# 5. Fit on TRAIN only
model.fit(X_train, y_train)
print("Features after encoding:", model.named_steps['prep'].get_feature_names_out().shape[0])
```

```python
# 6. Evaluate on the test set
proba = model.predict_proba(X_test)[:, 1]
pred = model.predict(X_test)

print("Random-guess PR-AUC:", round(y_test.mean(), 3))
print("PR-AUC:   ", round(average_precision_score(y_test, proba), 3))
print("ROC-AUC:  ", round(roc_auc_score(y_test, proba), 3))
print("Recall:   ", round(recall_score(y_test, pred), 3))
print("Precision:", round(precision_score(y_test, pred), 3))
print(classification_report(y_test, pred))
```
